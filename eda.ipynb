```python
trans_df["date"] = pd.to_datetime(trans_df["date"], format="%y%m%d")
trans_df["year_month"] = trans_df["date"].dt.to_period("M")
```

```python
trans_df["k_symbol"] = trans_df["k_symbol"].astype(str).str.strip()
trans_df["k_symbol"] = trans_df["k_symbol"].replace({"nan": np.nan, "": np.nan})
```

```python
trans_df["credit_amount"] = np.where(trans_df["type"] == "PRIJEM", trans_df["amount"], 0)
trans_df["debit_amount"]  = np.where(trans_df["type"] != "PRIJEM", trans_df["amount"], 0)
```

```python
# one-hot encoding for k_symbol

ksym_flags = pd.get_dummies(trans_df["k_symbol"], prefix="ksym")
trans_df = pd.concat([trans_df, ksym_flags], axis=1)
```

```python
agg_dict = {
    "credit_amount": "sum",
    "debit_amount": "sum",
    "trans_id": "count",
    "balance": "last",
}
agg_dict.update({col: "sum" for col in ksym_flags.columns})

monthly_panel = (
    trans_df.sort_values("date")
    .groupby(["account_id", "year_month"])
    .agg(agg_dict)
    .rename(columns={"trans_id": "n_transactions", "balance": "end_of_month_balance"})
    .reset_index()
)

monthly_panel["net_flow"] = monthly_panel["credit_amount"] - monthly_panel["debit_amount"]
```

```python
monthly_panel.head(10)
```

```python
monthly_panel = monthly_panel.sort_values(["account_id", "year_month"]).reset_index(drop=True)

WINDOW = 6      # kaç aylık geçmişe bakılacak
MIN_HISTORY_MONTHS = 6    # baseline için en az kaç ay geçmiş şart


## ortalama ve standart sapma

for col in ["credit_amount", "debit_amount", "end_of_month_balance"]:
    grp = monthly_panel.groupby("account_id")[col]
    monthly_panel[f"{col}_roll_mean"] = grp.transform(
        lambda s: s.shift(1).rolling(WINDOW, min_periods=MIN_HISTORY_MONTHS).mean()
    )
    monthly_panel[f"{col}_roll_std"] = grp.transform(
        lambda s: s.shift(1).rolling(WINDOW, min_periods=MIN_HISTORY_MONTHS).std()
    )


## z-score

for col in ["credit_amount", "debit_amount", "end_of_month_balance"]:
    monthly_panel[f"{col}_zscore"] = (
        (monthly_panel[col] - monthly_panel[f"{col}_roll_mean"])
        / monthly_panel[f"{col}_roll_std"]
    )

zscore_cols = [c for c in monthly_panel.columns if c.endswith("_zscore")]
monthly_panel[zscore_cols] = monthly_panel[zscore_cols].replace([np.inf, -np.inf], np.nan)
```

```python
# mutlak sapmayı da hesapla
monthly_panel["debit_amount_abs_diff"] = (
    monthly_panel["debit_amount"] - monthly_panel["debit_amount_roll_mean"]
)

Z_THRESHOLD = 2.0
MIN_ABS_CHANGE = 1000  # bu tutarın altındaki sapmalar iş açısından önemsiz sayılır

monthly_panel["debit_spike_flag"] = (
    (monthly_panel["debit_amount_zscore"].abs() >= Z_THRESHOLD) &
    (monthly_panel["debit_amount_abs_diff"].abs() >= MIN_ABS_CHANGE)
)
```

```python
# her hesabın kaçıncı ayında olduğunu işaretle (0 = ilk ay)
monthly_panel["months_of_history"] = monthly_panel.groupby("account_id").cumcount()
```

```python
# her hesabın kaçıncı ayında olduğunu işaretle (0 = ilk ay)
monthly_panel["months_of_history"] = monthly_panel.groupby("account_id").cumcount()

def compute_spike_flag(df, col, z_threshold=2.0, min_abs_change=1000, min_history=4):
    """
    Bir sütun için tetikleyici bayrağı üretir.
    Üç şart birden sağlanmalı:
      1) istatistiksel olarak sapmalı  (|z-skor| >= z_threshold)
      2) mutlak olarak da anlamlı      (|fark| >= min_abs_change)
      3) hesap yeterince olgun          (months_of_history >= min_history)
    """
    abs_diff = df[col] - df[f"{col}_roll_mean"]
    flag = (
        (df[f"{col}_zscore"].abs() >= z_threshold) &
        (abs_diff.abs() >= min_abs_change) &
        (df["months_of_history"] >= min_history)
    )
    return flag, abs_diff

# iki sütun için de spike
monthly_panel["debit_spike_flag"], monthly_panel["debit_amount_abs_diff"] = \
    compute_spike_flag(monthly_panel, "debit_amount", min_abs_change=1000, min_history=MIN_HISTORY_MONTHS)

monthly_panel["credit_spike_flag"], monthly_panel["credit_amount_abs_diff"] = \
    compute_spike_flag(monthly_panel, "credit_amount", min_abs_change=1500, min_history=MIN_HISTORY_MONTHS)
```

```python
monthly_panel["debit_direction"] = np.sign(monthly_panel["debit_amount_zscore"]).map({1: "up", -1: "down", 0: "flat"})
monthly_panel["credit_direction"] = np.sign(monthly_panel["credit_amount_zscore"]).map({1: "up", -1: "down", 0: "flat"})
```

```python
## Bu ayki bakiye, kendi son 12 aylık geçmişimin yüzde kaçından daha yüksek
```

```python
def rolling_percentile_rank(series, window=12, min_periods=6):
    def _pct_rank(x):
        if len(x) < min_periods:
            return np.nan
        current = x.iloc[-1]
        hist = x.iloc[:-1]
        return (hist < current).mean()
    return series.rolling(window, min_periods=min_periods).apply(_pct_rank, raw=False)

monthly_panel["balance_percentile"] = (
    monthly_panel.groupby("account_id")["end_of_month_balance"]
    .transform(lambda s: rolling_percentile_rank(s))
)
```

```python
# eğer bu ayki harcama önceki ayların %80 inden fazlaysa yüksek harcama diye etiketledi, %20 sinden düşükse düşük harcama
# sustained -> 3 aydır üst üste azalma veya artma varsa sustained diye etiketliyoruz
```

```python
HIGH_PCTL = 0.80
LOW_PCTL = 0.20

monthly_panel["high_balance_month"] = monthly_panel["balance_percentile"] >= HIGH_PCTL
monthly_panel["low_balance_month"]  = monthly_panel["balance_percentile"] <= LOW_PCTL

monthly_panel["sustained_high_balance_flag"] = (
    monthly_panel.groupby("account_id")["high_balance_month"]
    .transform(lambda s: s.rolling(3, min_periods=3).sum() == 3)
)
monthly_panel["sustained_low_balance_flag"] = (
    monthly_panel.groupby("account_id")["low_balance_month"]
    .transform(lambda s: s.rolling(3, min_periods=3).sum() == 3)
)
```

```python
## k_symbol tetikleyicileri

## daha önce hiç yapılmayan bir ödeme türü (fatura ödemesi gibi) yokken bu ay ortaya çıkarsa "onset" tetikleyicisi
## veya ilk defa negatif bakiye ceza faizi (SANKC. UROK) -> onset

## sürekli olan bi ödeme bi anda hiç yapılmamaya başlarsa "cessation" (kesilme) tetikleyicisi
```

```python
K_SYMBOL_COLS = [c for c in monthly_panel.columns if c.startswith("ksym_")]

trigger_records = []

for col in K_SYMBOL_COLS:
    symbol_name = col.replace("ksym_", "")
    present = (monthly_panel[col] > 0)

    seen_before = (
        monthly_panel.groupby("account_id")[col]
        .transform(lambda s: (s > 0).cumsum().shift(1, fill_value=0) > 0)
    )

    onset_flag = present & ~seen_before
    cessation_flag = (~present) & seen_before & (
        monthly_panel.groupby("account_id")[col]
        .transform(lambda s: (s > 0).shift(1, fill_value=False))
    )

    monthly_panel[f"{symbol_name}_onset_flag"] = onset_flag
    monthly_panel[f"{symbol_name}_cessation_flag"] = cessation_flag
```

```python
for col in K_SYMBOL_COLS:
    symbol_name = col.replace("ksym_", "")
    onset_rate = monthly_panel[f"{symbol_name}_onset_flag"].mean()
    cessation_rate = monthly_panel[f"{symbol_name}_cessation_flag"].mean()
    print(f"{symbol_name:15s} onset={onset_rate:.3f}  cessation={cessation_rate:.3f}")
```

```python
## SIPO onset -> yeni düzenlik yükümlülük
## SANKC. UROK onset -> ilk kez negatife düşme
## UVER cessation -> kredi bitmii olabilir (yeni kredi teklif fırsatı)
```

## TETİKLEYİCİLER

```python
## Confidence nasıl hesaplanıyor — kaynağa göre farklı mantık:

## Sayısal spike'larda z-skorun büyüklüğünden türetiyoruz (0-1 aralığına sıkıştırarak)
## Sustained balance'ta persentilin kendisini kullanıyoruz (0.95 persentil = 0.95 güven)
## Onset/cessation'da bunlar zaten kategorik/kesin olaylar olduğu için sabit ama olayın niteliğine göre farklılaştırılmış bir güven değeri veriyoruz
```

```python
records = []

# --- 1) Sayısal sapma tetikleyicileri: debit / credit spike ---
numeric_triggers = [
    {
        "flag_col": "debit_spike_flag", "zscore_col": "debit_amount_zscore",
        "direction_col": "debit_direction",
        "label_up": "unusual_spending_increase", "product_up": "ihtiyac_kredisi_taksitlendirme",
        "label_down": "unusual_spending_decrease", "product_down": None,
    },
    {
        "flag_col": "credit_spike_flag", "zscore_col": "credit_amount_zscore",
        "direction_col": "credit_direction",
        "label_up": "income_increase", "product_up": "yatirim_urunu_limit_artirimi",
        "label_down": "income_decrease", "product_down": None,
    },
]

for spec in numeric_triggers:
    sub = monthly_panel[monthly_panel[spec["flag_col"]]].copy()
    if len(sub) == 0:
        continue
    sub["trigger_type"] = np.where(sub[spec["direction_col"]] == "up", spec["label_up"], spec["label_down"])
    sub["suggested_product"] = np.where(sub[spec["direction_col"]] == "up", spec["product_up"], spec["product_down"])
    sub["confidence"] = sub[spec["zscore_col"]].abs().clip(upper=6) / 6
    sub["evidence"] = sub[spec["zscore_col"]].round(2).apply(lambda z: {"zscore": z})
    records.append(sub[["account_id", "year_month", "trigger_type", "confidence", "suggested_product", "evidence"]])

# --- 2) Kalıcı bakiye durumu tetikleyicileri ---
balance_triggers = [
    {"flag_col": "sustained_high_balance_flag", "label": "idle_cash_buildup",
     "product": "vadeli_mevduat_fon", "conf_from_pctl": True},
    {"flag_col": "sustained_low_balance_flag", "label": "liquidity_pressure",
     "product": "esnek_limit_kredi_guvence", "conf_from_pctl": False},
]

for spec in balance_triggers:
    sub = monthly_panel[monthly_panel[spec["flag_col"]]].copy()
    if len(sub) == 0:
        continue
    sub["trigger_type"] = spec["label"]
    sub["suggested_product"] = spec["product"]
    sub["confidence"] = sub["balance_percentile"] if spec["conf_from_pctl"] else 1 - sub["balance_percentile"]
    sub["evidence"] = sub["balance_percentile"].round(2).apply(lambda p: {"balance_percentile": p})
    records.append(sub[["account_id", "year_month", "trigger_type", "confidence", "suggested_product", "evidence"]])

# --- 3) k_symbol tabanlı onset / cessation tetikleyicileri ---
meaningful_onset = {
    "SIPO": ("new_recurring_bill", "esnek_odeme_bildirim_servisi", 0.75),
    "SANKC. UROK": ("first_overdraft", "ek_limit_kredi_guvence", 0.90),
    "DUCHOD": ("pension_income_started", "emekli_bankaciligi_paketi", 0.85),
}
meaningful_cessation = {
    "UVER": ("loan_paid_off", "yeni_kredi_teklifi", 0.80),
}

for symbol, (label, product, conf) in meaningful_onset.items():
    col = f"{symbol}_onset_flag"
    sub = monthly_panel[monthly_panel[col]].copy()
    if len(sub) == 0:
        continue
    sub["trigger_type"] = label
    sub["suggested_product"] = product
    sub["confidence"] = conf
    sub["evidence"] = [{"k_symbol": symbol, "event": "onset"}] * len(sub)
    records.append(sub[["account_id", "year_month", "trigger_type", "confidence", "suggested_product", "evidence"]])

for symbol, (label, product, conf) in meaningful_cessation.items():
    col = f"{symbol}_cessation_flag"
    sub = monthly_panel[monthly_panel[col]].copy()
    if len(sub) == 0:
        continue
    sub["trigger_type"] = label
    sub["suggested_product"] = product
    sub["confidence"] = conf
    sub["evidence"] = [{"k_symbol": symbol, "event": "cessation"}] * len(sub)
    records.append(sub[["account_id", "year_month", "trigger_type", "confidence", "suggested_product", "evidence"]])

# --- Birleştir ---
trigger_table = pd.concat(records, ignore_index=True)
trigger_table = trigger_table.sort_values(["account_id", "year_month"]).reset_index(drop=True)
```
